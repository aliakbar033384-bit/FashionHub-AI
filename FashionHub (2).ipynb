```python
import pandas as pd

# Load merged dataset (you already created this earlier)
merged_df = pd.read_csv("FashionHub_merged.csv")

# ---- AI Stylist: Filter Based ---- #
def ai_stylist(gender=None, price_range=None, occasion=None, season=None, n=5):
    df = merged_df.copy()

    # Filter by gender
    if gender:
        df = df[df["Gender"].str.lower() == gender.lower()]

    # Filter by season
    if season:
        df = df[df["Season"].str.lower() == season.lower()]

    # Filter by occasion (AI_Tags contains multiple tags, so we use "contains")
    if occasion:
        df = df[df["AI_Tags"].str.contains(occasion, case=False, na=False)]

    # Filter by price range
    if price_range == "low":
        df = df[df["PricePKR"] < 5000]
    elif price_range == "mid":
        df = df[(df["PricePKR"] >= 5000) & (df["PricePKR"] <= 12000)]
    elif price_range == "high":
        df = df[df["PricePKR"] > 12000]

    # Random sample of recommendations
    if len(df) > n:
        df = df.sample(n)

    return df[["ProductName", "Category", "SubCategory", "PricePKR", "Gender", "Season", "AI_Tags", "SellerName_x"]]


# ---- Example Usage ---- #
print("🎯 Example 1: Women's Summer Casual Outfit (Mid Price Range)")
print(ai_stylist(gender="Women", price_range="mid", occasion="casual", season="Summer"))

print("\n🎯 Example 2: Men's Wedding Outfit (High Price Range)")
print(ai_stylist(gender="Men", price_range="high", occasion="wedding", season="Winter"))
```

```python
import pandas as pd

# Load products dataset
products_df = pd.read_csv("FashionHub_products_updated.csv")

# Function to generate tags
def generate_tags(row):
    tags = []

    # Occasion
    if "Wedding" in row["SubCategory"] or "Shalwar Kameez" in row["SubCategory"] or "Saree" in row["SubCategory"]:
        tags.append("wedding")
    elif row["Category"] in ["Western Wear", "Eastern Wear"]:
        tags.append("casual")
    if "Suit" in row["SubCategory"] or "Blazer" in row["SubCategory"]:
        tags.append("formal")
    if "T-Shirt" in row["SubCategory"] or "Hoodie" in row["SubCategory"]:
        tags.append("trendy")

    # Season
    if row["Season"] in ["Summer", "Winter", "All-season"]:
        tags.append(row["Season"].lower())

    # Style
    if row["Category"] == "Eastern Wear":
        tags.append("ethnic")
    elif row["Category"] == "Western Wear":
        tags.append("western")
    else:
        tags.append("modern")

    # Gender
    tags.append(row["Gender"].lower())

    # Material
    if row["Material"].lower() in ["cotton", "silk", "lawn", "wool", "denim", "polyester", "leather", "linen"]:
        tags.append(row["Material"].lower())

    return ", ".join(sorted(set(tags)))  # remove duplicates + keep tidy

# Apply function
products_df["AI_Tags_Cleaned"] = products_df.apply(generate_tags, axis=1)

# Save updated dataset
products_df.to_csv("FashionHub_products_with_tags.csv", index=False)

# Quick check
print(products_df[["ProductName", "Category", "SubCategory", "Material", "Season", "Gender", "AI_Tags_Cleaned"]].head(10))
```

```python
import pandas as pd

# Load your products dataset
df = pd.read_csv("FashionHub_products_updated.csv")

# --- Tagging function ---
def generate_tags(row):
    tags = []

    # Gender-based
    if pd.notna(row["Gender"]):
        tags.append(row["Gender"].lower())

    # Season-based
    if pd.notna(row["Season"]):
        tags.append(row["Season"].lower())

    # Material-based
    if "leather" in str(row["Material"]).lower():
        tags.append("leather")
    if "wool" in str(row["Material"]).lower():
        tags.append("wool")
    if "linen" in str(row["Material"]).lower():
        tags.append("linen")
    if "silk" in str(row["Material"]).lower():
        tags.append("silk")
    if "lawn" in str(row["Material"]).lower():
        tags.append("lawn")
    if "polyester" in str(row["Material"]).lower():
        tags.append("polyester")

    # Category/Subcategory rules
    if row["Category"] == "Eastern Wear":
        tags.extend(["ethnic", "traditional"])
    if row["Category"] == "Western Wear":
        tags.append("western")
    if row["Category"] == "Jewelry":
        tags.append("trendy")
    if row["Category"] == "Bags & Accessories":
        tags.append("modern")

    # General styles
    if "dress" in str(row["SubCategory"]).lower() or "kameez" in str(row["SubCategory"]).lower():
        tags.append("casual")
    if "hoodie" in str(row["SubCategory"]).lower():
        tags.append("trendy")
    if "saree" in str(row["SubCategory"]).lower():
        tags.append("formal")

    # Remove duplicates
    return ", ".join(sorted(set(tags)))


# Apply function
df["AI_Tags_Cleaned"] = df.apply(generate_tags, axis=1)

# Save new dataset
df.to_csv("FashionHub_products_tagged.csv", index=False)

print("✅ File saved as FashionHub_products_tagged.csv with cleaned tags.")
```

```python
import pandas as pd

# Load your cleaned product dataset
df = pd.read_csv("FashionHub_products_updated.csv")

# --- Step 1: Define the cleaning function ---
def clean_ai_tags(tags):
    if pd.isna(tags):
        return ""
    
    # Split by comma, strip spaces, make lowercase
    tag_list = [t.strip().lower() for t in tags.split(",")]
    
    # Remove duplicates while keeping order
    seen = set()
    cleaned = []
    for tag in tag_list:
        if tag not in seen and tag != "":
            cleaned.append(tag)
            seen.add(tag)
    
    # Join back
    return ", ".join(cleaned)

# --- Step 2: Apply cleaning ---
df["AI_Tags_Cleaned"] = df["AI_Tags"].apply(clean_ai_tags)

# --- Step 3: Random sample of 20 products for validation ---
sample_df = df.sample(20, random_state=42)

# --- Step 4: Display only important columns ---
print(sample_df[['ProductID','SellerName','Category','SubCategory',
                 'ProductName','PricePKR','Gender','Material','Season',
                 'AI_Tags','AI_Tags_Cleaned']])
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("FashionHub_products_updated.csv")

# --- Rule-based Gender Correction ---
def fix_gender(row):
    subcat = row['SubCategory'].lower()
    gender = row['Gender']

    # Eastern Wear rules
    if subcat in ['saree', 'lehenga', 'kurti', 'abaya', 'dupatta']:
        return 'Women'
    # Jewelry rules (usually unisex, but not kids-only)
    elif subcat in ['necklace', 'ring', 'bracelet', 'earrings']:
        return 'Unisex'
    # Bags & Accessories
    elif subcat in ['handbag', 'wallet', 'belt']:
        return 'Unisex'
    # Shoes (sandals, heels, sneakers)
    elif subcat in ['sandals', 'heels']:
        return 'Women'
    elif subcat in ['sneakers', 'shoes']:
        return 'Men'
    # Western Wear rules
    elif subcat in ['t-shirt', 'jeans', 'hoodie', 'maxi dress']:
        if subcat == 'maxi dress':
            return 'Women'
        else:
            return 'Men' if gender == 'Men' else 'Unisex'
    # Default keep existing
    else:
        return gender

# Apply corrections
df['Corrected_Gender'] = df.apply(fix_gender, axis=1)

# Save corrected dataset
df.to_csv("FashionHub_products_corrected.csv", index=False)

# Show sample of corrected records
print(df[['ProductID','SellerName','Category','SubCategory','ProductName','Gender','Corrected_Gender']].sample(20, random_state=42))
```

```python
import pandas as pd

# Load the cleaned product dataset
df = pd.read_csv("FashionHub_products_updated.csv")

def ai_stylist(gender=None, season=None, occasion=None, price_range=None, n=5):
    """
    Recommend products based on AI stylist logic.
    
    Parameters:
    - gender: 'Men', 'Women', 'Kids', 'Unisex' (string)
    - season: 'Summer', 'Winter', 'All-season' (string)
    - occasion: e.g., 'wedding', 'casual', 'party', 'formal' (string)
    - price_range: tuple(min_price, max_price)
    - n: number of products to recommend
    
    Returns:
    - DataFrame of recommended products
    """
    df_filtered = df.copy()
    
    # Gender filter
    if gender:
        df_filtered = df_filtered[df_filtered['Gender'].str.lower() == gender.lower()]
    
    # Season filter
    if season:
        df_filtered = df_filtered[df_filtered['Season'].str.lower() == season.lower()]
    
    # Occasion filter (check if occasion tag exists in AI_Tags_Cleaned)
    if occasion:
        df_filtered = df_filtered[df_filtered['AI_Tags'].str.contains(occasion, case=False, na=False)]
    
    # Price range filter
    if price_range:
        min_price, max_price = price_range
        df_filtered = df_filtered[(df_filtered['PricePKR'] >= min_price) & (df_filtered['PricePKR'] <= max_price)]
    
    # Randomize and select top N products
    recommendations = df_filtered.sample(n=min(n, len(df_filtered)), random_state=42)
    
    return recommendations[['ProductID','SellerName','Category','SubCategory','ProductName','PricePKR','Gender','Material','Season','AI_Tags']]

# Example usage
# Recommend 5 products for Women in Summer for casual occasion priced 1000-15000 PKR
sample_recommendations = ai_stylist(gender='Women', season='Summer', occasion='casual', price_range=(1000,15000), n=5)

print(sample_recommendations)
```

```python
# ==============================
# FashionHub Phase 2: AI Stylist (Strict Reality-Based)
# ==============================

# ----- 1. Import Libraries -----
import pandas as pd

# ----- 2. Load Cleaned Dataset -----
df = pd.read_csv("FashionHub_products_cleaned.csv")

# ----- 3. Column Cleaning (Optional if already cleaned) -----
def clean_columns(df):
    for col in ['Category', 'SubCategory', 'Material', 'Season', 'Gender']:
        df[col] = df[col].astype(str).str.strip().str.title()
    material_map = {
        'Cottn': 'Cotton',
        'Silkk': 'Silk',
        'Polyester ': 'Polyester',
        'Leather ': 'Leather',
        'Wool ': 'Wool',
        'Denim ': 'Denim',
        'Linen ': 'Linen'
    }
    df['Material'] = df['Material'].replace(material_map)
    return df

df = clean_columns(df)

# ----- 4. Gender-Category Mapping for Strict Reality Check -----
gender_category_map = {
    'Men': ['Western Wear', 'Shoes', 'Bags & Accessories', 'Others'],
    'Women': ['Eastern Wear', 'Western Wear', 'Shoes', 'Jewelry', 'Bags & Accessories'],
    'Unisex': ['Kids Wear', 'Shoes', 'Accessories', 'Others'],
    'Kids': ['Kids Wear', 'Shoes', 'Others']
}

def filter_by_gender_category_strict(df, gender):
    allowed_categories = gender_category_map.get(gender, [])
    return df[df['Category'].isin(allowed_categories)]

# ----- 5. Phase 2 Strict AI Stylist Function -----
def ai_stylist_phase2_strict(gender=None, season=None, occasion=None, price_range=None, n=5):
    """
    Phase 2 AI Stylist with strict reality-based filtering.
    """
    df_filtered = df.copy()

    # Filter by AI_Tags_Cleaned (must match all selected tags)
    if gender:
        df_filtered = df_filtered[df_filtered['AI_Tags_Cleaned'].str.contains(gender, case=False, na=False)]
    if season:
        df_filtered = df_filtered[df_filtered['AI_Tags_Cleaned'].str.contains(season, case=False, na=False)]
    if occasion:
        df_filtered = df_filtered[df_filtered['AI_Tags_Cleaned'].str.contains(occasion, case=False, na=False)]

    # Filter by price
    if price_range:
        min_price, max_price = price_range
        df_filtered = df_filtered[(df_filtered['PricePKR'] >= min_price) & (df_filtered['PricePKR'] <= max_price)]

    # Strict reality-based filter: gender-category enforcement
    if gender:
        df_filtered = filter_by_gender_category_strict(df_filtered, gender)

    # Randomize and select top N
    recommendations = df_filtered.sample(n=min(n, len(df_filtered)), random_state=42)

    return recommendations[['ProductID','SellerName','Category','SubCategory','ProductName','PricePKR','Gender','Material','Season','AI_Tags_Cleaned']]

# ----- 6. Example Usage -----
# Women, Summer, casual, 1000-15000 PKR, top 5
sample_recommendations_1 = ai_stylist_phase2_strict(
    gender='Women',
    season='Summer',
    occasion='casual',
    price_range=(1000, 15000),
    n=5
)
print("---- Sample Recommendations 1 ----")
print(sample_recommendations_1)

# Men, Winter, formal, 5000-20000 PKR, top 5
sample_recommendations_2 = ai_stylist_phase2_strict(
    gender='Men',
    season='Winter',
    occasion='formal',
    price_range=(5000, 20000),
    n=5
)
print("\n---- Sample Recommendations 2 ----")
print(sample_recommendations_2)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("FashionHub_products_cleaned.csv")

# Allowed categories per gender
gender_category_map = {
    'Men': ['Western Wear', 'Eastern Wear', 'Shoes', 'Bags & Accessories', 'Others'],
    'Women': ['Western Wear', 'Eastern Wear', 'Shoes', 'Bags & Accessories', 'Jewelry', 'Others'],
    'Unisex': ['Western Wear', 'Eastern Wear', 'Shoes', 'Bags & Accessories', 'Others']
}

# Unrealistic subcategory and product keywords per gender
gender_exclusion_map = {
    'Men': {
        'SubCategory': ['Saree', 'Abaya', 'Kids Wear', 'Necklace', 'Earrings', 'Handbag'],
        'Keywords': ['saree', 'abaya', 'necklace', 'earrings', 'handbag', 'kids']
    },
    'Women': {
        'SubCategory': ['Kids Wear', 'Tie', 'Wallet', 'Cap', 'Belt', 'Shoes (Men)'],
        'Keywords': ['tie', 'wallet', 'cap', 'belt', 'mens', 'boys', 'kids']
    },
    'Unisex': {
        'SubCategory': [],
        'Keywords': []
    }
}

def ai_stylist_advanced(gender=None, season=None, occasion=None, price_range=None, n=5):
    df_filtered = df.copy()

    if gender:
        gender_lower = gender.lower()
        df_filtered = df_filtered[df_filtered['Corrected_Gender'].str.lower() == gender_lower]

        # Allowed categories
        allowed_categories = gender_category_map.get(gender, [])
        df_filtered = df_filtered[df_filtered['Category'].isin(allowed_categories)]

        # Exclude unrealistic subcategories
        exclusions = gender_exclusion_map.get(gender, {})
        subcat_excl = exclusions.get('SubCategory', [])
        keyword_excl = exclusions.get('Keywords', [])

        df_filtered = df_filtered[~df_filtered['SubCategory'].isin(subcat_excl)]
        # Also exclude by product keywords
        for kw in keyword_excl:
            df_filtered = df_filtered[~df_filtered['ProductName'].str.lower().str.contains(kw)]

    # Filter by season
    if season:
        df_filtered = df_filtered[df_filtered['Season'].str.lower() == season.lower()]

    # Filter by occasion
    if occasion:
        df_filtered = df_filtered[df_filtered['AI_Tags_Cleaned'].str.contains(occasion, case=False, na=False)]

    # Filter by price
    if price_range:
        min_price, max_price = price_range
        df_filtered = df_filtered[(df_filtered['PricePKR'] >= min_price) & (df_filtered['PricePKR'] <= max_price)]

    # Randomize and select top N
    recommendations = df_filtered.sample(n=min(n, len(df_filtered)), random_state=42)

    return recommendations[['ProductID','SellerName','Category','SubCategory','ProductName','PricePKR','Corrected_Gender','Material','Season','AI_Tags_Cleaned']]

# Test examples
women_recs = ai_stylist_advanced(gender='Women', season='Summer', occasion='casual', price_range=(1000,15000), n=5)
men_recs = ai_stylist_advanced(gender='Men', season='Winter', occasion='formal', price_range=(1000,15000), n=5)

print("---- Women Recommendations ----")
print(women_recs)
print("\n---- Men Recommendations ----")
print(men_recs)
```

```python
def ai_outfit_stylist(starting_product_id, n_matches=1):
    """
    Recommend a matching outfit based on a starting product.
    
    Parameters:
    - starting_product_id: ProductID to base the outfit on
    - n_matches: number of matching items per category
    
    Returns:
    - DataFrame of recommended outfit items
    """
    starting_item = df[df['ProductID'] == starting_product_id]
    if starting_item.empty:
        return pd.DataFrame()
    
    category = starting_item.iloc[0]['Category']
    subcat = starting_item.iloc[0]['SubCategory']
    gender = starting_item.iloc[0]['Corrected_Gender']
    season = starting_item.iloc[0]['Season']
    
    # Determine matching categories
    matching_map = {
        'Western Wear': ['Shoes', 'Bags & Accessories', 'Jewelry'],
        'Eastern Wear': ['Shoes', 'Bags & Accessories', 'Jewelry'],
        'Shoes': ['Western Wear', 'Eastern Wear', 'Bags & Accessories'],
        'Bags & Accessories': ['Western Wear', 'Eastern Wear', 'Shoes'],
        'Jewelry': ['Eastern Wear', 'Western Wear', 'Shoes']
    }
    
    match_cats = matching_map.get(category, [])
    
    outfit_items = [starting_item]
    
    for cat in match_cats:
        matches = df[
            (df['Category'] == cat) &
            (df['Corrected_Gender'] == gender) &
            (df['Season'] == season)
        ].sample(n=min(n_matches, len(df)), random_state=42)
        
        outfit_items.append(matches)
    
    return pd.concat(outfit_items)[['ProductID','SellerName','Category','SubCategory','ProductName','PricePKR','Corrected_Gender','Material','Season','AI_Tags_Cleaned']]

# Example: Generate outfit for a Saree (Women)
outfit = ai_outfit_stylist(starting_product_id='P0814', n_matches=1)
print(outfit)
```

```python
import pandas as pd

# Load the cleaned dataset
df = pd.read_csv("FashionHub_products_cleaned.csv")

# Define matching categories map
matching_map = {
    'Eastern Wear': ['Shoes', 'Bags & Accessories', 'Jewelry'],
    'Western Wear': ['Shoes', 'Bags & Accessories', 'Jewelry'],
    'Shoes': ['Eastern Wear', 'Western Wear', 'Bags & Accessories'],
    'Bags & Accessories': ['Eastern Wear', 'Western Wear', 'Shoes'],
    'Jewelry': ['Eastern Wear', 'Western Wear', 'Shoes']
}

def ai_outfit_builder(start_product_id):
    # Get the starting product
    start_product = df[df['ProductID'] == start_product_id].iloc[0]
    
    gender = start_product['Corrected_Gender']
    season = start_product['Season']
    category = start_product['Category']
    
    outfit = pd.DataFrame([start_product])  # Start with main product
    
    # Find matching categories
    if category in matching_map:
        for match_cat in matching_map[category]:
            # Filter products by matching category, same gender, same season
            matches = df[
                (df['Category'] == match_cat) &
                (df['Corrected_Gender'] == gender) &
                (df['Season'] == season) &
                (df['Category'].str.lower() != 'kids wear')  # skip kids wear
            ]
            if not matches.empty:
                # Pick one matching product randomly
                outfit = pd.concat([outfit, matches.sample(1, random_state=42)])
    
    return outfit[['ProductID', 'ProductName', 'Category', 'SubCategory', 'Corrected_Gender', 'Material', 'Season', 'AI_Tags_Cleaned']]

# ----- Example usage ----- #
# Pick a product, e.g., a Saree
example_outfit = ai_outfit_builder('P0814')
print(example_outfit)
```

```python
import pandas as pd

# Load the cleaned product dataset
df = pd.read_csv("FashionHub_products_cleaned.csv")

def ai_stylist_filtered(gender, season=None, occasion=None, price_range=None, n=5, sort_by=None, trending=False):
    """
    Recommend products based on user-selected gender and optional filters.
    
    Parameters:
    - gender: 'Women', 'Men', 'Kids', 'Unisex'
    - season: 'Summer', 'Winter', 'All-season' (optional)
    - occasion: e.g., 'wedding', 'casual', 'party', 'formal' (optional)
    - price_range: tuple(min_price, max_price) (optional)
    - n: number of products to recommend
    - sort_by: 'price_low_to_high', 'price_high_to_low', 'latest', None (optional)
    - trending: bool, if True prioritize trending/popular items (optional)
    
    Returns:
    - DataFrame of recommended products
    """
    df_filtered = df[df['Corrected_Gender'].str.lower() == gender.lower()]

    # Skip 'Kids Wear' for Men/Women
    if gender.lower() in ['men', 'women']:
        df_filtered = df_filtered[~df_filtered['Category'].str.lower().eq('kids wear')]

    # Filter by season
    if season:
        df_filtered = df_filtered[df_filtered['Season'].str.lower() == season.lower()]

    # Filter by occasion
    if occasion:
        df_filtered = df_filtered[df_filtered['AI_Tags_Cleaned'].str.contains(occasion, case=False, na=False)]

    # Filter by price range
    if price_range:
        min_price, max_price = price_range
        df_filtered = df_filtered[(df_filtered['PricePKR'] >= min_price) & (df_filtered['PricePKR'] <= max_price)]

    # Apply trending/popularity filter if available
    if trending and 'PopularityScore' in df_filtered.columns:
        df_filtered = df_filtered.sort_values('PopularityScore', ascending=False)
    else:
        # Sorting options
        if sort_by == 'price_low_to_high':
            df_filtered = df_filtered.sort_values('PricePKR', ascending=True)
        elif sort_by == 'price_high_to_low':
            df_filtered = df_filtered.sort_values('PricePKR', ascending=False)
        elif sort_by == 'latest':
            df_filtered = df_filtered.sort_values('ProductID', ascending=False)
        else:
            df_filtered = df_filtered.sample(frac=1, random_state=42)  # random if no sort

    # Pick top N
    recommendations = df_filtered.head(n)

    return recommendations[['ProductID','SellerName','Category','SubCategory',
                            'ProductName','PricePKR','Corrected_Gender','Material','Season','AI_Tags_Cleaned']]

# --- Example Usage ---
sample_recommendations = ai_stylist_filtered(
    gender='Women', 
    season='Summer', 
    occasion='casual', 
    price_range=(1000,15000), 
    n=5,
    sort_by='price_low_to_high',
    trending=True
)
print(sample_recommendations)
```

```python
import pandas as pd

class FashionHubRecommender:
    def __init__(self, product_csv_path):
        # Load product dataset
        self.df = pd.read_csv(product_csv_path)

        # Clean AI tags column (handle missing values)
        self.df['AI_Tags_Cleaned'] = self.df['AI_Tags_Cleaned'].fillna('').astype(str)

    def recommend(self, gender=None, season=None, selected_tags=None, top_n=10):
        """
        Recommend products based on gender, season, and AI tags.

        Parameters:
        - gender: str ("Men", "Women", "Unisex") or None
        - season: str ("Summer", "Winter", etc.) or None
        - selected_tags: list of tags (["casual", "party", "ethnic"]) or None
        - top_n: int (number of results to return)

        Returns:
        - DataFrame of top recommended products
        """
        df_filtered = self.df.copy()

        # --- Apply filters ---
        if gender:
            df_filtered = df_filtered[df_filtered['Gender'].str.lower() == gender.lower()]

        if season:
            df_filtered = df_filtered[df_filtered['Season'].str.lower() == season.lower()]

        # --- Scoring function ---
        def tag_score(product_tags):
            if not selected_tags:
                return 0
            product_tags_set = set([t.strip().lower() for t in product_tags.split(',') if t.strip()])
            return len(product_tags_set.intersection(set([t.lower() for t in selected_tags])))

        df_filtered['Score'] = df_filtered['AI_Tags_Cleaned'].apply(tag_score)

        # --- Sort by score (highest first) ---
        df_sorted = df_filtered.sort_values(by='Score', ascending=False)

        # --- Return top N products ---
        return df_sorted.head(top_n)[
            ['ProductID', 'ProductName', 'PricePKR', 'SellerName', 'AI_Tags_Cleaned']
        ]

# ----------------- Example Usage -----------------
if __name__ == "__main__":
    recommender = FashionHubRecommender("FashionHub_products_cleaned.csv")

    # Example: Recommend summer casual outfits for women
    recommended = recommender.recommend(
        gender="Women",
        season="Summer",
        selected_tags=["casual", "party"],
        top_n=20
    )

    print("\nRecommended Products:")
    print(recommended.to_string(index=False))
```

```python
import pandas as pd
import random

# Load your dataset
df = pd.read_csv("FashionHub_products_cleaned.csv")

# -------------------------
# Category Mapping
# -------------------------
category_map = {
    "tops": ["shirt", "kurta", "kameez", "blouse", "hoodie", "t-shirt", "abaya", "saree"],
    "bottoms": ["jeans", "trousers", "shalwar", "pants", "skirt"],
    "footwear": ["shoes", "heels", "sandals", "sneakers"],
    "accessories": ["scarf", "dupatta", "belt", "bracelet", "ring", "hat", "cap", "jewelry"],
    "bags": ["handbag", "backpack", "clutch"],
    "outerwear": ["jacket", "coat", "sweater"]
}

def detect_category(product_name):
    name = product_name.lower()
    for cat, keywords in category_map.items():
        if any(word in name for word in keywords):
            return cat
    return "other"

df["Category"] = df["ProductName"].apply(detect_category)

# -------------------------
# Styling Rules
# -------------------------
style_rules = {
    "abaya": ["accessories", "footwear", "bags"],      # scarf + flats + handbag
    "saree": ["footwear", "accessories", "bags"],      # heels + jewelry + clutch
    "jeans": ["tops", "footwear", "bags"],             # shirt + sneakers + backpack
    "shirt": ["bottoms", "footwear", "accessories"],   # trousers + shoes + belt
}

# -------------------------
# Outfit Builder Function
# -------------------------
def build_outfit(selected_product_id, df):
    main_item = df[df["ProductID"] == selected_product_id].iloc[0]
    main_name = main_item["ProductName"].lower()
    main_cat = main_item["Category"]
    tags = set(main_item["AI_Tags_Cleaned"].split(", "))
    season = main_item["Season"]
    gender = main_item["Gender"]

    # Outfit starter
    outfit = [main_item]

    # Check which style rule applies
    matched_rule = None
    for keyword, comp_cats in style_rules.items():
        if keyword in main_name:
            matched_rule = comp_cats
            break

    if matched_rule:
        for comp_cat in matched_rule:
            candidates = df[(df["Category"] == comp_cat) &
                            (df["Season"] == season) &
                            (df["Gender"] == gender)]

            # Match tags
            candidates = candidates[candidates["AI_Tags_Cleaned"].apply(
                lambda x: len(tags.intersection(set(x.split(", ")))) > 0
            )]

            if not candidates.empty:
                outfit.append(candidates.sample(1).iloc[0])

    return pd.DataFrame(outfit)

# -------------------------
# DEMO: Generate Logical Outfits
# -------------------------
print("=== Abaya Outfit ===")
abaya_outfit = build_outfit("P0089", df)  # Pink Lawn Abaya
print(abaya_outfit[["ProductID", "ProductName", "Category", "AI_Tags_Cleaned"]])

print("\n=== Saree Outfit ===")
saree_outfit = build_outfit("P0814", df)  # Green Lawn Saree
print(saree_outfit[["ProductID", "ProductName", "Category", "AI_Tags_Cleaned"]])

print("\n=== Jeans Outfit ===")
jeans_outfit = build_outfit("P0192", df)  # Beige Denim Jeans
print(jeans_outfit[["ProductID", "ProductName", "Category", "AI_Tags_Cleaned"]])
```

```python
import pandas as pd
import random

# ========================
# 1. Load Dataset
# ========================
df = pd.read_csv("FashionHub_products_cleaned.csv")

# Add dummy popularity + feedback (for now)
df["Popularity"] = df["PricePKR"] % 100   # proxy: price variation
df["Feedback"] = (df.index % 5) * 10      # proxy: likes/wishlist

# ========================
# 2. Category Mapping
# ========================
category_map = {
    "Shirt": "Top", "T-Shirt": "Top", "Kurta": "Top", "Blouse": "Top", "Hoodie": "Top",
    "Abaya": "Dress/Modest", "Maxi": "Dress/Modest", "Saree": "Dress/Modest", "Gown": "Dress/Modest",
    "Trouser": "Bottom", "Jeans": "Bottom", "Skirt": "Bottom", "Shalwar": "Bottom", "Churidar": "Bottom",
    "Dupatta": "Accessory", "Scarf": "Accessory", "Ring": "Accessory", "Necklace": "Accessory", "Bag": "Accessory",
    "Shoes": "Shoes", "Heels": "Shoes", "Sneakers": "Shoes", "Sandals": "Shoes"
}

def map_category(name):
    for key, value in category_map.items():
        if key.lower() in name.lower():
            return value
    return "Other"

df["MappedCategory"] = df["ProductName"].apply(map_category)

# ========================
# 3. Occasion Rules
# ========================
occasion_rules = {
    "casual": {"must_have": ["Top", "Bottom"], "optional": ["Shoes", "Accessory"]},
    "formal": {"must_have": ["Top", "Bottom"], "optional": ["Shoes", "Accessory"]},
    "party": {"must_have": ["Dress/Modest"], "optional": ["Shoes", "Accessory"]},
    "wedding": {"must_have": ["Dress/Modest"], "optional": ["Shoes", "Accessory", "Accessory"]}
}

# Icons for display
category_icons = {
    "Top": "👕", "Bottom": "👖", "Shoes": "👠", "Accessory": "👜", "Dress/Modest": "👗", "Other": "✨"
}

# ========================
# 4. Outfit Builder Around a Product
# ========================
def build_outfit_around(product_id, top_n=3):
    main_item = df[df["ProductID"] == product_id]
    if main_item.empty:
        print(f"❌ Product {product_id} not found.")
        return []

    main_item = main_item.iloc[0]
    occasion = None
    for key in occasion_rules.keys():
        if key in str(main_item["AI_Tags_Cleaned"]).lower():
            occasion = key
            break

    if not occasion:
        occasion = "casual"  # fallback

    rules = occasion_rules[occasion]
    must_have, optional = rules.get("must_have", []), rules.get("optional", [])

    outfits = []
    for _ in range(10):  # generate multiple outfit candidates
        outfit = [main_item]

        def filter_items(category):
            items = df[df["MappedCategory"] == category]
            if main_item["Season"]:
                items = items[items["Season"].str.contains(main_item["Season"], case=False, na=False)]
            if main_item["Gender"]:
                items = items[items["Gender"].str.contains(main_item["Gender"], case=False, na=False)]
            return items

        # Must-have (excluding main item category)
        for cat in must_have:
            if cat != main_item["MappedCategory"]:
                items = filter_items(cat)
                if not items.empty:
                    outfit.append(items.sample(1).iloc[0])

        # Optional
        for cat in optional:
            if random.random() > 0.5:
                items = filter_items(cat)
                if not items.empty:
                    outfit.append(items.sample(1).iloc[0])

        if outfit:
            outfit_df = pd.DataFrame(outfit)

            # Scoring
            outfit_df["ItemScore"] = (
                3 + outfit_df["Popularity"] * 0.5 + outfit_df["Feedback"] * 0.8
            )
            outfit_score = outfit_df["ItemScore"].sum()
            outfits.append((outfit_score, outfit_df))

    outfits = sorted(outfits, key=lambda x: x[0], reverse=True)
    return [(score, outfit) for score, outfit in outfits[:top_n]]

# ========================
# 5. Display Outfits
# ========================
def show_outfits_for_product(product_id):
    outfits = build_outfit_around(product_id)
    if not outfits:
        return
    main_item = df[df["ProductID"] == product_id].iloc[0]
    print(f"\n=== Outfit Suggestions for {main_item['ProductName']} ({main_item['ProductID']}) ===")
    for score, outfit in outfits:
        print(f"\nOutfit Score: {int(score)}")
        for _, row in outfit.iterrows():
            cat = row["MappedCategory"]
            icon = category_icons.get(cat, "✨")
            print(f"{icon} {cat}: {row['ProductName']}")
        print("-" * 40)

# ========================
# 6. Test Example
# ========================
show_outfits_for_product("P0089")  # Example: Pink Lawn Abaya
```

```python
import pandas as pd
import random

# -------------------------------
# 1. Sample Product Dataset
# -------------------------------
products = [
    {"ProductID": "P001", "ProductName": "Blue Denim Jeans", "Category": "Bottom", "Occasion": "Casual", "Season": "All"},
    {"ProductID": "P002", "ProductName": "White Cotton Shirt", "Category": "Top", "Occasion": "Casual", "Season": "Summer"},
    {"ProductID": "P003", "ProductName": "Black Formal Trousers", "Category": "Bottom", "Occasion": "Formal", "Season": "All"},
    {"ProductID": "P004", "ProductName": "Grey Blazer", "Category": "Outerwear", "Occasion": "Formal", "Season": "Winter"},
    {"ProductID": "P005", "ProductName": "Red Lawn Kurti", "Category": "Top", "Occasion": "Casual", "Season": "Summer"},
    {"ProductID": "P006", "ProductName": "Pink Lawn Abaya", "Category": "Dress", "Occasion": "Traditional", "Season": "Summer"},
    {"ProductID": "P007", "ProductName": "Beige Wool Shawl", "Category": "Outerwear", "Occasion": "Traditional", "Season": "Winter"},
    {"ProductID": "P008", "ProductName": "Green Cotton Hoodie", "Category": "Outerwear", "Occasion": "Casual", "Season": "Winter"},
    {"ProductID": "P009", "ProductName": "Leather Formal Shoes", "Category": "Shoes", "Occasion": "Formal", "Season": "All"},
    {"ProductID": "P010", "ProductName": "White Sneakers", "Category": "Shoes", "Occasion": "Casual", "Season": "All"},
]
df = pd.DataFrame(products)

# -------------------------------
# 2. Outfit Rules (Logical Mapping)
# -------------------------------
outfit_rules = {
    "Casual": {
        "Top": ["Bottom", "Shoes", "Outerwear"],
        "Bottom": ["Top", "Shoes"],
        "Outerwear": ["Top", "Bottom", "Shoes"]
    },
    "Formal": {
        "Top": ["Bottom", "Outerwear", "Shoes"],
        "Bottom": ["Top", "Outerwear", "Shoes"],
        "Outerwear": ["Top", "Bottom", "Shoes"]
    },
    "Traditional": {
        "Dress": ["Shoes", "Outerwear"],
        "Top": ["Bottom", "Shoes", "Outerwear"],
        "Outerwear": ["Dress", "Top", "Shoes"]
    }
}

# -------------------------------
# 3. Outfit Builder Function
# -------------------------------
def build_outfit(occasion="Casual", season=None):
    """
    Generate a logical outfit based on occasion + season.
    """
    candidates = df[df["Occasion"] == occasion]

    if season:
        candidates = candidates[(candidates["Season"] == season) | (candidates["Season"] == "All")]

    if candidates.empty:
        return f"No outfits found for {occasion} in {season}"

    outfit = []
    used_categories = set()

    # Start with a random base item (like Top or Dress)
    base_item = candidates.sample(1).iloc[0]
    outfit.append(base_item)
    used_categories.add(base_item["Category"])

    # Match other categories based on rules
    if base_item["Category"] in outfit_rules[occasion]:
        valid_matches = outfit_rules[occasion][base_item["Category"]]

        for category in valid_matches:
            options = candidates[candidates["Category"] == category]
            if not options.empty and category not in used_categories:
                match = options.sample(1).iloc[0]
                outfit.append(match)
                used_categories.add(match["Category"])

    return pd.DataFrame(outfit)

# -------------------------------
# 4. Demo
# -------------------------------
print("=== Casual Summer Outfit ===")
print(build_outfit(occasion="Casual", season="Summer"))

print("\n=== Formal Winter Outfit ===")
print(build_outfit(occasion="Formal", season="Winter"))

print("\n=== Traditional Winter Outfit ===")
print(build_outfit(occasion="Traditional", season="Winter"))
```

```python
# premium_ai_stylist.py
# Paste into Jupyter or save & run with: python premium_ai_stylist.py
import pandas as pd
import numpy as np
import random
from typing import List, Dict

random.seed(42)
np.random.seed(42)

# ---------- 0. Load dataset ----------
# Update the path if needed. Assumes file is in same folder.
CSV_PATH = "FashionHub_products_cleaned.csv"

df = pd.read_csv(CSV_PATH)

# ---------- 1. Make sure required columns exist ----------
# We saw the user's dataset columns earlier:
# ['ProductID', 'SellerID', 'SellerName', 'Category', 'SubCategory',
#  'ProductName', 'PricePKR', 'Gender', 'Color', 'Material', 'Season',
#  'StockQty', 'AI_Tags', 'Corrected_Gender', 'AI_Tags_Cleaned']

# Normalize some column names that might vary
if "AI_Tags_Cleaned" not in df.columns and "AI_Tags" in df.columns:
    df["AI_Tags_Cleaned"] = df["AI_Tags"].astype(str)
if "AI_Tags_Cleaned" not in df.columns:
    df["AI_Tags_Cleaned"] = df["ProductName"].astype(str)  # fallback

# add Popularity and Feedback columns if missing (dummy values)
if "Popularity" not in df.columns:
    df["Popularity"] = np.random.randint(50, 1000, size=len(df))
if "Feedback" not in df.columns:
    df["Feedback"] = np.round(np.random.uniform(3.0, 5.0, size=len(df)), 1)

# Ensure Category column exists and normalized
if "Category" not in df.columns:
    # try SubCategory fallback or infer from ProductName
    if "SubCategory" in df.columns:
        df["Category"] = df["SubCategory"]
    else:
        df["Category"] = df["ProductName"].apply(lambda x: "Other")

# normalize text columns
df["AI_Tags_Cleaned"] = df["AI_Tags_Cleaned"].fillna("").astype(str).str.lower()
df["Category"] = df["Category"].fillna("Other").astype(str)
df["Season"] = df["Season"].fillna("all").astype(str).str.lower()
df["Gender"] = df["Gender"].fillna("").astype(str).str.lower()

# ---------- 2. Category keyword mapping (helps when Category names are messy) ----------
# Keywords mapped to normalized category types we use for outfit rules
KEYWORD_CATEGORY_MAP = {
    "jeans": "bottom",
    "trouser": "bottom",
    "trousers": "bottom",
    "pant": "bottom",
    "shalwar": "bottom",
    "kurta": "top",
    "kameez": "top",
    "shirt": "top",
    "t-shirt": "top",
    "tshirt": "top",
    "hoodie": "top",
    "sweater": "outerwear",
    "jacket": "outerwear",
    "coat": "outerwear",
    "abaya": "dress_modest",
    "saree": "dress_modest",
    "gown": "dress_modest",
    "maxi": "dress_modest",
    "lehenga": "dress_modest",
    "dress": "dress_modest",
    "kurti": "top",
    "dupatt": "dupatt a",  # will match dupatta with startswith check
    "dupatta": "dupatt a",
    "scarf": "accessory",
    "belt": "accessory",
    "ring": "accessory",
    "necklace": "accessory",
    "bracelet": "accessory",
    "bag": "bag",
    "handbag": "bag",
    "backpack": "bag",
    "clutch": "bag",
    "shoe": "shoes",
    "shoes": "shoes",
    "sneaker": "shoes",
    "sneakers": "shoes",
    "heels": "shoes",
    "sandals": "shoes",
    "khussa": "shoes",
    "loafers": "shoes",
}

def normalize_category(name: str) -> str:
    name_l = str(name).lower()
    # direct match against existing category string
    for k, normalized in KEYWORD_CATEGORY_MAP.items():
        if k in name_l:
            return normalized
    # fallback: use original category (lower)
    return name_l.strip()

df["NormCategory"] = df["Category"].apply(normalize_category)

# ---------- 3. Gender-aware outfit rule sets ----------
# We'll keep separate templates for male & female, plus neutral fallback
OUTFIT_TEMPLATES = {
    "female": {
        "casual": ["top", "bottom", "shoes", "accessory"],
        "formal": ["top", "bottom", "shoes", "accessory"],
        "party": ["dress_modest", "shoes", "accessory", "bag"],
        "wedding": ["dress_modest", "shoes", "accessory", "bag"],
        "ethnic": ["top", "bottom", "dupatt a", "shoes", "accessory"],
        "work": ["top", "bottom", "shoes", "accessory"]
    },
    "male": {
        "casual": ["top", "bottom", "shoes", "outerwear"],
        "formal": ["top", "bottom", "shoes", "outerwear"],
        "party": ["top", "bottom", "shoes", "accessory"],
        "wedding": ["dress_modest", "shoes", "accessory"],  # dress_modest for sherwani etc.
        "ethnic": ["top", "bottom", "shoes", "accessory"],
        "work": ["top", "bottom", "shoes"]
    },
    "neutral": {
        "casual": ["top", "bottom", "shoes"],
        "formal": ["top", "bottom", "shoes"],
        "party": ["dress_modest", "shoes", "accessory"],
        "wedding": ["dress_modest", "shoes", "accessory"],
        "ethnic": ["top", "bottom", "dupatt a", "shoes"],
        "work": ["top", "bottom", "shoes"]
    }
}

# ---------- 4. Scoring helpers ----------
def relevance_tag_score(item_tags: str, desired_tags: List[str]) -> int:
    # desired_tags: e.g., ['summer', 'ethnic', 'party']
    if not item_tags:
        return 0
    item_set = set([t.strip() for t in item_tags.split(",") if t.strip()])
    # count matches
    return sum(1 for dt in desired_tags if dt in item_set)

def item_score(row: pd.Series, desired_tags: List[str]) -> float:
    # base relevance (tag matches)
    rel = relevance_tag_score(row["AI_Tags_Cleaned"], desired_tags)
    # popularity normalized and feedback weighted
    pop = row.get("Popularity", 0)
    fb = row.get("Feedback", 0)
    # normalize pop roughly to 0-1 by dividing by a scale (we'll use 1000)
    pop_norm = pop / 1000.0
    # feedback is typically 1-5 -> normalize to 0-1
    fb_norm = (fb - 1.0) / 4.0 if fb >= 1 else 0
    # final weighted score
    return rel * 2.0 + pop_norm * 1.5 + fb_norm * 1.0

# ---------- 5. Outfit generation core ----------
def generate_style_aware_outfits(
    occasion: str,
    season: str,
    style_preference: str,
    gender: str = None,
    n_outfits: int = 3,
    top_k_candidates: int = 10
) -> List[Dict]:
    """
    Returns a list of outfits. Each outfit is a dict:
    { "score": float, "items": [row_dicts...] }
    Inputs:
     - occasion: casual|formal|party|wedding|ethnic|work
     - season: summer|winter|all
     - style_preference: western|ethnic|trendy|modest (string matched against AI_Tags_Cleaned)
     - gender: 'male' | 'female' or None
    """
    occ = str(occasion).lower()
    sea = str(season).lower()
    style = str(style_preference).lower()
    gender_norm = (str(gender).lower() if gender else None)
    if gender_norm not in ("male", "female"):
        gender_key = "neutral"
    else:
        gender_key = gender_norm

    # choose template
    template = OUTFIT_TEMPLATES.get(gender_key, OUTFIT_TEMPLATES["neutral"])
    if occ not in template:
        occ = "casual"  # fallback
    slot_list = template[occ]  # sequence of category keys to fill

    # desired tags list used for relevance
    desired_tags = [t.strip().lower() for t in [sea, style, occ] if t]

    # Pre-filter the catalog by season & style & (optionally) gender-ish tags
    def product_filter(df_local):
        # season: allow "all" or matching season
        df_local = df_local[df_local["Season"].fillna("all").str.lower().isin([sea, "all"])]
        # style_preference: must appear in AI tags (if style empty skip)
        if style:
            df_local = df_local[df_local["AI_Tags_Cleaned"].str.contains(style, na=False)]
        # optional: prefer gender-specific matches if Gender column exists and user specified gender
        if gender_norm:
            # match Corrected_Gender or Gender if exists
            if "Corrected_Gender" in df_local.columns:
                df_local = df_local[df_local["Corrected_Gender"].fillna("").str.lower().str.contains(gender_norm) | df_local["Corrected_Gender"].isnull()]
            elif "Gender" in df_local.columns:
                df_local = df_local[df_local["Gender"].fillna("").str.lower().str.contains(gender_norm) | df_local["Gender"].isnull()]
        return df_local

    catalog = df.copy()
    catalog = product_filter(catalog)

    if catalog.empty:
        return []

    # For each category slot, build candidate pools
    # We map normalized category keys (slot_list entries) to products whose NormCategory matches.
    candidates_by_slot = {}
    for slot in set(slot_list):
        # slot might be e.g. "dupatt a" or "dress_modest", etc.
        pool = catalog[catalog["NormCategory"].str.contains(slot, na=False)]
        # fallback: if pool empty, try matching by Category exact lowercase
        if pool.empty:
            pool = catalog[catalog["Category"].str.lower() == slot]
        # compute item scores and keep top_k_candidates
        if not pool.empty:
            pool = pool.copy()
            pool["ItemScore"] = pool.apply(lambda r: item_score(r, desired_tags), axis=1)
            pool = pool.sort_values(by="ItemScore", ascending=False).head(top_k_candidates)
        candidates_by_slot[slot] = pool

    # Generate outfit candidates by sampling from candidate pools (yield up to many tries)
    outfits = []
    attempts = 0
    max_attempts = max(n_outfits * 30, 100)  # try enough times to get varied outfits
    while len(outfits) < n_outfits and attempts < max_attempts:
        attempts += 1
        chosen_items = []
        valid = True
        for slot in slot_list:
            pool = candidates_by_slot.get(slot, pd.DataFrame())
            if pool.empty:
                # fallback: try any item in catalog that matches season & style & not duplicate category
                fallback_pool = catalog.copy()
                fallback_pool["ItemScore"] = fallback_pool.apply(lambda r: item_score(r, desired_tags), axis=1)
                fallback_pool = fallback_pool.sort_values("ItemScore", ascending=False).head(top_k_candidates)
                if fallback_pool.empty:
                    valid = False
                    break
                chosen = fallback_pool.sample(1).iloc[0]
            else:
                chosen = pool.sample(1).iloc[0]
            # avoid duplicate productID in same outfit
            if any(chosen["ProductID"] == it["ProductID"] for it in chosen_items):
                valid = False
                break
            chosen_items.append(chosen.to_dict())
        if not valid:
            continue
        # compute outfit score = sum item scores (recompute to be safe)
        total_score = sum(item_score(pd.Series(item), desired_tags) for item in chosen_items)
        outfits.append({"score": total_score, "items": chosen_items})

    # sort outfits by score descending and return
    outfits = sorted(outfits, key=lambda x: x["score"], reverse=True)
    return outfits

# ---------- 6. Pretty display helper ----------
def print_outfits(outfits: List[Dict], occasion: str, season: str, style_preference: str):
    if not outfits:
        print("No outfits found for these inputs.")
        return
    print(f"\n=== AI Stylist: {occasion.title()} | {season.title()} | {style_preference.title()} ===")
    for idx, outfit in enumerate(outfits, 1):
        print("\n" + ("=" * 48))
        print(f"Outfit {idx}  —  Score: {int(outfit['score'])}")
        for item in outfit["items"]:
            cat = item.get("Category", item.get("NormCategory", ""))
            print(f" • {cat.title():15} : {item.get('ProductName')}  |  PKR {item.get('PricePKR')}  |  Seller: {item.get('SellerName')}")
        print("=" * 48)

# ---------- 7. Example usage ----------
if __name__ == "__main__":
    # Example 1: Female, Casual, Summer, Ethnic style
    outfits1 = generate_style_aware_outfits(occasion="casual", season="summer", style_preference="ethnic", gender="female", n_outfits=3)
    print_outfits(outfits1, "casual", "summer", "ethnic")

    # Example 2: Female, Wedding, Summer, Ethnic
    outfits2 = generate_style_aware_outfits(occasion="wedding", season="summer", style_preference="ethnic", gender="female", n_outfits=3)
    print_outfits(outfits2, "wedding", "summer", "ethnic")

    # Example 3: Male, Party, Winter, Western
    outfits3 = generate_style_aware_outfits(occasion="party", season="winter", style_preference="western", gender="male", n_outfits=3)
    print_outfits(outfits3, "party", "winter", "western")
```

```python
from sklearn.preprocessing import MinMaxScaler

def calculate_score(df, season=None, trend_tags=[]):
    """
    Calculate a weighted score for products based on popularity, feedback, trends, and season.
    """

    # Normalize Popularity and Feedback
    scaler = MinMaxScaler()
    df[['Popularity_norm', 'Feedback_norm']] = scaler.fit_transform(df[['Popularity', 'Feedback']])

    # Trend Boost (1 if any AI tag matches trend_tags, else 0)
    df['TrendBoost'] = df['AI_Tags_Cleaned'].apply(
        lambda tags: 1 if any(tag in tags for tag in trend_tags) else 0
    )

    # Seasonal Boost (1 if product season matches outfit season, else 0)
    df['SeasonalBoost'] = df['Season'].apply(
        lambda s: 1 if season and s.lower() == season.lower() else 0
    )

    # Weighted scoring
    w1, w2, w3, w4 = 0.3, 0.4, 0.2, 0.1
    df['Score'] = (
        (w1 * df['Popularity_norm']) +
        (w2 * df['Feedback_norm']) +
        (w3 * df['TrendBoost']) +
        (w4 * df['SeasonalBoost'])
    )

    return df
```

```python
def safe_pick(df, category, top_n=5):
    """
    Safely pick one product from a category. 
    Returns empty DataFrame if none found.
    """
    subset = df[df['Category'] == category]
    if subset.empty:
        print(f"⚠️ No products found for category '{category}', skipping...")
        return pd.DataFrame()
    return subset.sort_values('Score', ascending=False).head(top_n).sample(1)


def generate_outfits(products, template, season=None, trend_tags=None, num_outfits=5):
    outfits = []
    products_scored = calculate_score(products.copy(), season=season, trend_tags=trend_tags)

    for outfit_num in range(1, num_outfits+1):
        # Core picks with safe fallback
        top = safe_pick(products_scored, "Top")
        bottom = safe_pick(products_scored, "Bottom")
        shoes = safe_pick(products_scored, "Shoes")
        accessory = safe_pick(products_scored, "Accessory")

        # Fallback if accessory missing → add extra Top
        if accessory.empty:
            accessory = safe_pick(products_scored, "Top")

        # Combine items
        for item in [top, bottom, shoes, accessory]:
            if not item.empty:
                outfits.append({
                    "Template": template,
                    "OutfitNumber": outfit_num,
                    **item.iloc[0].to_dict()
                })

    return pd.DataFrame(outfits)
```

```python
import pandas as pd
import random

# -------------------------
# Config / input files
# -------------------------
PRODUCTS_CSV = "FashionHub_products_cleaned.csv"  # update if path different
OUTPUT_CSV = "FashionHub_Outfits_WebReady_WithSimilar.csv"
BASE_URL = "https://www.fashionhub.pk/product/"

# -------------------------
# Load product dataset
# -------------------------
df = pd.read_csv(PRODUCTS_CSV)

# Provide defaults if some columns missing
if 'Score' not in df.columns:
    df['Score'] = 100
if 'PricePKR' not in df.columns:
    df['PricePKR'] = 0
if 'ProductID' not in df.columns:
    df['ProductID'] = ["P{:04d}".format(i) for i in range(len(df))]
if 'SellerName' not in df.columns:
    df['SellerName'] = "Unknown Seller"
if 'Stock' not in df.columns:
    df['Stock'] = 1
if 'Season' not in df.columns:
    df['Season'] = 'All-season'
if 'Gender' not in df.columns:
    # default to all genders if missing
    df['Gender'] = df.get('Gender', 'unisex')
if 'Popularity' not in df.columns:
    df['Popularity'] = 0
if 'Feedback' not in df.columns:
    df['Feedback'] = 0

# -------------------------
# Category keywords mapping
# -------------------------
category_keywords = {
    "Top": ["Shirt", "Kurta", "T-Shirt", "Hoodie", "Jacket", "Sweater", "Blouse", "Tunic"],
    "Bottom": ["Shalwar", "Jeans", "Trousers", "Saree", "Skirt", "Pant"],
    "Shoes": ["Shoes", "Sandals", "Sneakers", "Heels", "Flats", "Loafers", "Boots"],
    "Accessory": ["Bag", "Wallet", "Belt", "Scarf", "Necklace", "Ring", "Cap", "Handbag"]
}

shoes_priority = {
    "western": ["Sneakers", "Heels", "Flats", "Loafers", "Boots", "Sandals"],
    "ethnic": ["Sandals", "Heels", "Flats", "Loafers", "Boots", "Sneakers"]
}

# -------------------------
# Outfit generation function
# -------------------------
def generate_outfits(df, occasion, season, style, gender, n_outfits=10, rng_seed=None):
    if rng_seed is not None:
        random.seed(rng_seed)
    # filter by gender and season
    filtered = df[df['Gender'].str.lower().isin([gender.lower(), 'unisex', 'all'])]
    filtered = filtered[filtered['Season'].str.lower().isin([season.lower(), 'all-season'])]
    categories = ["Top", "Bottom", "Shoes", "Accessory"]
    outfits = []
    for i in range(n_outfits):
        outfit = {}
        for cat in categories:
            # find candidates by keywords
            kw = category_keywords.get(cat, [])
            if len(kw) == 0:
                candidates = filtered
            else:
                pattern = '|'.join(kw)
                candidates = filtered[filtered['ProductName'].str.contains(pattern, case=False, na=False)]
            # shoes style priority
            if cat == "Shoes" and style.lower() in shoes_priority:
                priority_list = shoes_priority[style.lower()]
                temp_candidates = candidates[candidates['ProductName'].str.contains('|'.join(priority_list), case=False, na=False)]
                if not temp_candidates.empty:
                    candidates = temp_candidates
            # fallback
            if candidates.empty:
                chosen = filtered.sample(n=1).iloc[0]
            else:
                top_candidates = candidates.sort_values("Score", ascending=False).head(5)
                chosen = top_candidates.sample(n=1).iloc[0]
            outfit[cat] = f"{chosen['ProductName']} (Season: {chosen['Season']}, Score: {chosen['Score']})"
        outfit["TemplateName"] = f"{occasion}_{season}_{style}_{gender}"
        outfit["OutfitNumber"] = i + 1
        outfits.append(outfit)
    return outfits

# -------------------------
# Expanded templates (keep genders men/women only)
# -------------------------
occasions = ["casual", "party", "formal", "wedding", "office", "travel"]
seasons = ["summer", "winter", "all-season"]
styles = ["western", "ethnic"]
genders = ["men", "women"]  # keep limited as requested

templates = []
for occ in occasions:
    for sea in seasons:
        for st in styles:
            for g in genders:
                templates.append({"occasion": occ, "season": sea, "style": st, "gender": g, "n_outfits": 8})

# -------------------------
# Generate outfits for all templates
# -------------------------
all_outfits = []
for tpl in templates:
    outfs = generate_outfits(df, tpl["occasion"], tpl["season"], tpl["style"], tpl["gender"], tpl["n_outfits"])
    all_outfits.extend(outfs)

outfits_df = pd.DataFrame(all_outfits)

# -------------------------
# Map product info (using main df)
# -------------------------
def map_product_info(product_str):
    product_name = product_str.split(" (Season:")[0].strip()
    match = df[df['ProductName'].str.lower() == product_name.lower()]
    if not match.empty:
        row = match.iloc[0]
        return pd.Series({
            "ID": row['ProductID'],
            "Price": row['PricePKR'],
            "Seller": row['SellerName'],
            "Stock": row.get("Stock", 1),
            "Season": row.get("Season", "All-season"),
            "Popularity": row.get("Popularity", 0),
            "Feedback": row.get("Feedback", 0),
            "URL": f"{BASE_URL}{row['ProductID']}"
        })
    else:
        return pd.Series({
            "ID": None, "Price": 0, "Seller": None, "Stock": 0,
            "Season": "All-season", "Popularity": 0, "Feedback": 0, "URL": None
        })

for cat in ["Top","Bottom","Shoes","Accessory"]:
    mapped = outfits_df[cat].apply(map_product_info)
    outfits_df[f"{cat}_ID"] = mapped["ID"]
    outfits_df[f"{cat}_Price"] = mapped["Price"]
    outfits_df[f"{cat}_Seller"] = mapped["Seller"]
    outfits_df[f"{cat}_Stock"] = mapped["Stock"]
    outfits_df[f"{cat}_Season"] = mapped["Season"]
    outfits_df[f"{cat}_Popularity"] = mapped["Popularity"]
    outfits_df[f"{cat}_Feedback"] = mapped["Feedback"]
    outfits_df[f"{cat}_URL"] = mapped["URL"]

# -------------------------
# Calculations: totals, availability, aggregates
# -------------------------
outfits_df["Total_Price"] = (
    outfits_df["Top_Price"].fillna(0) + outfits_df["Bottom_Price"].fillna(0) +
    outfits_df["Shoes_Price"].fillna(0) + outfits_df["Accessory_Price"].fillna(0)
)

outfits_df["Availability"] = outfits_df[["Top_Stock","Bottom_Stock","Shoes_Stock","Accessory_Stock"]].min(axis=1).apply(
    lambda x: "Available" if x>0 else "Out of stock"
)

outfits_df["Max_Popularity"] = outfits_df[["Top_Popularity","Bottom_Popularity","Shoes_Popularity","Accessory_Popularity"]].max(axis=1)
outfits_df["Avg_Feedback"] = outfits_df[["Top_Feedback","Bottom_Feedback","Shoes_Feedback","Accessory_Feedback"]].mean(axis=1).fillna(0)

# -------------------------
# AI Stylist Score
# -------------------------
def season_style_weight(template_name):
    # give highest weight to summer+western currently, others mid
    name = template_name.lower()
    if "summer" in name and "western" in name:
        return 100
    elif "wedding" in name or "party" in name:
        return 90
    else:
        return 60

outfits_df["AI_Stylist_Score"] = (
    outfits_df['Max_Popularity'] * 0.4 +
    outfits_df['Avg_Feedback'] * 20 * 0.4 +   # feedback scaled 0-5 -> 0-100 by *20
    outfits_df['TemplateName'].apply(season_style_weight) * 0.2
)

# -------------------------
# Trending flag (top 20%)
# -------------------------
score_threshold = outfits_df['AI_Stylist_Score'].quantile(0.8)
outfits_df['Trending'] = outfits_df['AI_Stylist_Score'].apply(lambda x: "Trending" if x >= score_threshold else "Regular")

# -------------------------
# Helper columns for website (Style, Template_Season)
# -------------------------
outfits_df['Style'] = outfits_df['TemplateName'].apply(lambda x: x.split('_')[2] if len(x.split('_'))>2 else "")
outfits_df['Template_Season'] = outfits_df['TemplateName'].apply(lambda x: x.split('_')[1] if len(x.split('_'))>1 else "")

# -------------------------
# Create a unique OutfitID and index for quick lookup
# -------------------------
outfits_df = outfits_df.reset_index(drop=True)
outfits_df["OutfitID"] = outfits_df.apply(lambda r: f"{r['TemplateName']}|{r['OutfitNumber']}", axis=1)

# -------------------------
# Similar_Outfits: same occasion+season+style, top 3 by AI_Stylist_Score, excluding itself
# -------------------------
# We'll parse TemplateName to get occasion/season/style part (first 3 parts)
def tpl_key(template_name):
    parts = template_name.split('_')
    # join occasion, season, style to be safe
    if len(parts) >= 3:
        return "_".join(parts[:3])
    return template_name

outfits_df["tpl_key"] = outfits_df["TemplateName"].apply(tpl_key)

# Prepare groups
groups = outfits_df.groupby("tpl_key")

similar_map = {}
for key, group in groups:
    # sort group by AI score desc
    sorted_group = group.sort_values("AI_Stylist_Score", ascending=False)
    ids = list(sorted_group["OutfitID"])
    # for each outfit in this group, take top 3 other IDs
    for idx, row in sorted_group.iterrows():
        this_id = row["OutfitID"]
        # candidates excluding itself
        others = [o for o in ids if o != this_id]
        top_similar = others[:3]  # if less than 3, shorter list
        similar_map[this_id] = top_similar

# Attach Similar_Outfits column
outfits_df["Similar_Outfits"] = outfits_df["OutfitID"].apply(lambda oid: similar_map.get(oid, []))

# -------------------------
# Reorder columns for final CSV
# -------------------------
columns_order = [
    "OutfitID", "TemplateName", "OutfitNumber",
    "Style", "Template_Season",
    "Top", "Top_ID", "Top_Price", "Top_Seller", "Top_Stock", "Top_Season", "Top_Popularity", "Top_Feedback", "Top_URL",
    "Bottom", "Bottom_ID", "Bottom_Price", "Bottom_Seller", "Bottom_Stock", "Bottom_Season", "Bottom_Popularity", "Bottom_Feedback", "Bottom_URL",
    "Shoes", "Shoes_ID", "Shoes_Price", "Shoes_Seller", "Shoes_Stock", "Shoes_Season", "Shoes_Popularity", "Shoes_Feedback", "Shoes_URL",
    "Accessory", "Accessory_ID", "Accessory_Price", "Accessory_Seller", "Accessory_Stock", "Accessory_Season", "Accessory_Popularity", "Accessory_Feedback", "Accessory_URL",
    "Total_Price", "Availability", "Is_Available" if "Is_Available" in outfits_df.columns else "Max_Popularity", 
    "Max_Popularity", "Avg_Feedback", "AI_Stylist_Score", "Trending", "Similar_Outfits"
]

# ensure columns present (avoid duplicate or missing)
final_columns = [c for c in columns_order if c in outfits_df.columns]

final_df = outfits_df[final_columns]

# -------------------------
# Export final CSV
# -------------------------
final_df.to_csv(OUTPUT_CSV, index=False)
print(f"✅ Exported final CSV with Similar_Outfits to '{OUTPUT_CSV}'")
print("\nTop 5 Trending Outfits (sample):")
print(final_df[final_df['Trending']=="Trending"].sort_values("AI_Stylist_Score", ascending=False).head(5)[
    ["OutfitID","TemplateName","OutfitNumber","AI_Stylist_Score","Total_Price","Availability","Similar_Outfits"]
])
```

```python
def calculate_ai_stylist_score(row, current_season, outfit_style, budget_range=None):
    """
    row: product row (with popularity, feedback, price, season, tags, seller type, etc.)
    current_season: 'Summer', 'Winter', etc.
    outfit_style: 'casual', 'party', 'formal', 'wedding'
    budget_range: tuple (min_price, max_price) or None
    """

    # --- Base score ---
    base_score = (row["Popularity"] * row["Feedback"]) / 10

    # --- Season boost ---
    if row["Season"].lower() == current_season.lower():
        season_boost = 15
    else:
        season_boost = -10

    # --- Style boost ---
    tags = str(row.get("AI_Tags_Cleaned", "")).lower()
    if outfit_style in tags:
        style_boost = 10
    else:
        style_boost = 0

    # --- Seller boost (equal trust for all) ---
    seller_boost = 5   # same for big brands, local, Instagram, TikTok, bazaar

    # --- Price balance ---
    price_balance = 0
    if budget_range:
        min_price, max_price = budget_range
        if min_price <= row["PricePKR"] <= max_price:
            price_balance = 5
        elif row["PricePKR"] > max_price * 1.5:  # way overpriced
            price_balance = -10

    # --- Final Score ---
    return base_score + season_boost + style_boost + seller_boost + price_balance
```
